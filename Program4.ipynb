```python
"""
Use word embeddings to improve prompts for Generative AI model.
Retrieve similar words using word embeddings. Use the similar words to enrich a GenAI prompt.
Use the AI model to generate responses for the original and enriched prompts.
Compare the outputs in terms of detail and relevance.
"""
```

```python
import gensim.downloader as api

# Load embeddings
print("Loading word embeddings...")
word_vectors = api.load("glove-wiki-gigaword-50")
print("Loaded successfully!")
```

```python
# Get similar words
def get_similar_words(word, topn=2):
    if word in word_vectors:
        return [w for w, sim in word_vectors.most_similar(word, topn=topn)]
    return []


# Enrich prompt
def enrich_prompt(prompt):
    words = prompt.lower().split()
    enriched = []

    for word in words:
        similar = get_similar_words(word)
        if similar:
            enriched.append(f"{word} ({', '.join(similar)})")
        else:
            enriched.append(word)

    return " ".join(enriched)

# Dummy AI model
def genai_response(prompt):
    prompt = prompt.lower()

    if "tired" in prompt or "exhausted" in prompt:
        return "Students seem disengaged. Use interactive teaching methods, real-life examples, and activities to improve interest."

    if "students" in prompt and "subject" in prompt:
        return "Focus on improving engagement by simplifying concepts and using visual aids."

    return "This is a general response."
```

```python
original_prompt = "Students from my class are tired of my subject"
enriched_prompt = enrich_prompt(original_prompt)

print("\nOriginal Prompt:", original_prompt)
print("Enriched Prompt:", enriched_prompt)

# Generate responses
original_response = genai_response(original_prompt)
enriched_response = genai_response(enriched_prompt)

print("\nOriginal Response:", original_response)
print("Enriched Response:", enriched_response)

# Compare
print("\nComparison:")
print("Original Length:", len(original_response.split()))
print("Enriched Length:", len(enriched_response.split()))
```

```python
# Import InferenceClient class to connect to Hugging Face cloud models and run them without downloading locally
from huggingface_hub import InferenceClient
import os
# Create a InferenceClient object with your Hugging Face API token to authenticate and send requests to models
client = InferenceClient(token=os.getenv("your_api_token"))


def enrich_prompt(prompt):
    response = client.chat.completions.create(
        model="meta-llama/Llama-3.1-8B-Instruct",
        messages=[
            {
                "role": "user",
                "content": f"""Rewrite this prompt in a more detailed, clear, and effective way: {prompt}"""
            }
        ],
        max_tokens=100
    )
    return response.choices[0].message.content



def genai_response(prompt):
    response = client.chat.completions.create(
        model="meta-llama/Llama-3.1-8B-Instruct",
        messages=[
            {"role": "user", "content": prompt}
        ],
        max_tokens=150
    )
    return response.choices[0].message.content
```

```python
original_prompt = "write about Jyothy Institute of Technology"

# Step 1: Enhance prompt
enhanced_prompt = enrich_prompt(original_prompt)

print("\nOriginal Prompt:", original_prompt)
print("Enhanced Prompt:", enhanced_prompt)

# Step 2: Generate responses
original_response = genai_response(original_prompt)
enhanced_response = genai_response(enhanced_prompt)

print("\nOriginal Response:\n", original_response)
print("\nEnhanced Response:\n", enhanced_response)

# Step 3: Compare
print("\nComparison:")
print("Original Length:", len(original_response.split()))
print("Enhanced Length:", len(enhanced_response.split()))
```
