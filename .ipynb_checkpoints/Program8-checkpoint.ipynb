```python
from langchain_community.chat_models import ChatCohere
from langchain_community.document_loaders import TextLoader
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
import os
print(os.path.abspath("sample.txt"))
```

```python
loader = TextLoader(r"C:\Users\prajw\Gen AI\sample.txt", encoding="utf-8")
documents = loader.load()
text = documents[0].page_content
text
```

```python
prompt = ChatPromptTemplate.from_template(
    "Summarize this text clearly in 4-5 lines:\n\n{text}\n\nSummary:"
)
```

```python
import os
#from langchain_cohere import ChatCohere

# Ensure your environment variable name matches exactly (e.g., 'cohere_api')
llm = ChatCohere(cohere_api_key=os.getenv('cohere_api'), model="command-r-08-2024", temperature=0.7)
```

```python
chain = prompt | llm | StrOutputParser()
```

```python
result = chain.invoke({"text": text})

print("\nSummary:\n", result)
```
