```python
from transformers import pipeline
import torch
"""
1. Load model
2. Load tokenizer
3. Convert text → tokens (preprocessing)
4. Run model (inference)
5. Convert output → probabilities (softmax)
6. Convert probabilities → labels (postprocessing)

"""
# Load the sentiment analysis pipeline
classifier = pipeline(
    "sentiment-analysis", 
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
# Example sentences
sentences = [
"I love this product! It's amazing.",
"This is the worst experience I've ever had.",
"The food was okay, nothing special.",
"I'm feeling very happy today.",
"This movie was absolutely terrible."
]
```

```python
result = classifier(sentences)
print(result)
```

```python
# Analyze the sentiment of each sentence
for sentence in sentences:
  result = classifier(sentence)
  print(f"Sentence: {sentence}")
  print(result)
  print(f"Sentiment: {result[0]['label']}, Score: {result[0]['score']}")
  print("-" * 20)
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import torch.nn.functional as F

# Step 1: Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased-finetuned-sst-2-english"
)

# Step 2: Load model
model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased-finetuned-sst-2-english"
)

# Step 3: Input text
text = "This movie was absolutely terrible."

# Step 4: Convert text → tokens (tokenization)
inputs = tokenizer(text, return_tensors="pt")

# input_ids, attention_mask,
# Step 5: Run model (inference)
outputs = model(**inputs)


# Step 6: Convert logits → probabilities
probs = F.softmax(outputs.logits, dim=1)

# Step 7: Convert probabilities → label
predicted_class_id = torch.argmax(probs).item()

labels = ["NEGATIVE", "POSITIVE"]
predicted_label = labels[predicted_class_id]

print("Prediction:", predicted_label)
print("Confidence:", probs[0][predicted_class_id].item())
```
